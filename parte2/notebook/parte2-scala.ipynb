## Ejercicio 5 — Comparación de dos manos

Se crea una función que recibe dos puntuaciones y devuelve la mayor utilizando una estructura `if / else`.


```scala
def maxHand(a: Int, b: Int): Int = {
    if (a>b){
        a
    } else {
        b
    }
}

println(maxHand(17,19))
println(maxHand(20,18))
println(maxHand(21,21))
```

##### Información
Cuando ambos tienen el mismo valor se cumple la parte de `else` porque al comprobar si a>b va a ser false, por ende imprime el `Int` que representa el b
Considero que no hace falta modificar la función porque en caso de ser iguales, da igual si imprime el a o el b

## Ejercicio 6 — Decidir el ganador de una partida

La función `ganador` utiliza la función `bust` para comprobar si alguna de las manos supera 21 y devuelve la mejor puntuación.

```scala
def ganador(handA: Int, handB: Int): Int = {
    if (bust(handA) && bust(handB)) {
        0
    }
    else if (bust(handA)) {
        handB
    } else if (bust(handB)) {
        handA
    } else if (handA > handB) {
        handA
    } else {
        handB
    }
}

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
println(ganador(15,15))
```

##### Información
`26 y 20`: cumple en el primer else if, ya que `handA` supera 21, y se deveulve `handB`
`18 y 22`: cumple en el segundo else if, ya que `handB` supera 21, y se deveulve `handA`
`24 y 25`: se devuelve 0 ya que los dos superan 21
`17 y 19`: cumple en el ultimo else if, ya que ni `handA` ni `handB` supera 21, y se deveulve `handB`, el mayor de los dos
`21, 20`: ninguno supera 21 asi que se devuelve `handA`, el mayor
`15 y 15`: aunque son iguales, ninguno de los cumple asi que devuelve 15.

## Ejercicio 9 — Recorrer un Array con `while`

Recorrido a un array de puntuaciones utilizando un contador mutable y la función `bust`.

```scala
val manos = Array(17, 24, 21, 19, 26)

var i = 0

while (i< manos.length){
    println(s"Mano = ${manos(i)} | Bust= ${bust(manos(i))}")
    i = i + 1
}
```

##### Información
 El contador debe ser mutable porque cada vez que pasa por el bucle `while` lo modificaremos, sumando en este caso 1. En caso de que no se incremente el contador dentro del bucle, se quedaría atrapado dentro de él ya que nunca superaria la longitud del array.
El uso de un contador, su modificación y el control manual del recorrido son características del estilo imperativo.

## Ejercicio 12 — Operadores relacionales y lógicos

Uso de operadores relacionales y lógicos para comparar las puntuaciones de tres manos.

```scala
val handA = 18
val handB = 21
val handC = 25

println(handA>handB)
println(handB==21)
println(handC!=21)
println((handA <= 21) && (handB <= 21))
println((handA> 21) || (handC> 21))
println(!(handB > 21))
```

### Información

| Expresión | Resultado | Explicación |
|---|---|---|
| `handA > handB` | `false` | 18 no es mayor que 21 |
| `handB == 21` | `true` | handB vale 21 |
| `handC != 21` | `true` | 25 es diferente de 21 |
| `handA <= 21 && handB <= 21` | `true` | Ambas manos son menores o iguales a 21 |
| `handA > 21 || handC > 21` | `true` | handC supera 21 |
| `!(handB > 21)` | `true` | handB no supera 21 |

## Ejercicio 13 — `foreach` y funciones como valores

Comparación del recorrido de un array utilizando un bucle `while` o utilizando `foreach`.

```scala
val manos = Array(17, 24, 21, 26, 18)

def bust (hand: Int):Boolean = {
    hand> 21
}

var i = 0

while(i < manos.length){
   println(s"${manos(i)} -> ${bust(manos(i))}")
    i += 1
}
println("\n For each:")

manos.foreach(mano => { println(s"$mano -> ${bust(mano)}")
                      })



```

### Información

La versión con `while` es la que necesita un contador y una variable mutable para recorrer el array.

La versión con `foreach` no necesita controlar manualmente un índice ni modificar una variable durante el recorrido.

Por este motivo, la versión con `foreach` se aproxima más al estilo funcional.

## Ejercicio 14 — Efectos secundarios y estilo de programación

Se comparan dos funciones: una que modifica una variable externa y otra que únicamente recibe valores y devuelve un resultado.

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(19)
println(s"Total: $total")
sumarAlTotal(5)
println(s"Total: $total")
sumarAlTotal(2)
println(s"Total: $total")

def sumar(a: Int, b: Int): Int = {
  a + b
}

println(s"Sumar 2 y 3: ${sumar(2,3)}")
println(s"Sumar 2 y 3: ${sumar(2,3)}")
```

### Información 
`sumarAlTotal` modifica una variable situada fuera de la función. Esta modificación es un efecto secundario, porque altera una variable que no existe solo dentro de la función.
